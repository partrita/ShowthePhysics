## 데이터 저장 및 처리
데이터는 배열(array)에 '저장'할 수 있으며, 배열에 담긴 각 값은 처리하거나 그래프를 그리거나 출력하는 등 다양하게 활용할 수 있습니다. 아래 예시에는 $a$라는 배열이 있고, 몇 개의 값이 저장되어 있습니다. 이 값들을 출력한 뒤, 그중 하나의 값을 변경하고 다시 출력해 보겠습니다(오른쪽 상단의 {fa}`rocket` 아이콘을 클릭한 뒤 live code를 눌러 코드 셀을 실행할 수 있습니다).

```python
import numpy as np
a = np.array([0.0, 1.0, 1.2, 0.8, 0.5]) # 몇 개의 값으로 배열 만들기
print(a)                                # 값 출력하기
a[1] = 2.0                              # 두 번째 원소의 값 바꾸기
print(a)                                # 값 출력하기
```

위에서 볼 수 있듯이, 값을 화면에 출력하려면 `print` 명령을 사용합니다. 

데이터가 담긴 배열을 이용하면 몇 가지 통계 계산도 수행할 수 있습니다:

```python
a_mean = np.mean(a)                          # a의 평균값 계산
a_std = np.std(a,ddof=1)                     # a의 표준편차 계산
a_unc = np.std(a,ddof=1)/np.sqrt(len(a))     # a의 표준오차 계산

print("Mean value of a is: ",a_mean)         # a의 평균값 출력
print("Standard deviation of a is: ",a_std)  # a의 표준편차 출력
print("Standard error of a is: ",a_unc)      # a의 표준오차 출력

# 더 깔끔하게 출력하기:
print("Standard error of a is: ", round(a_unc,1))      # a의 표준오차 출력
```

```python
def sum_function(i1, i2): # 이 sum_function은 두 개의 입력 변수를 필요로 합니다
    temp_var = i1 + i2    # 두 입력 변수를 더해 임시 변수에 저장합니다
    return  temp_var      # 저장된 값을 반환합니다

x = sum_function(3, 5) 

print(x) # 8을 출력합니다
```

## 그래프 그리기

종종 데이터를 시각적으로 표현하고 싶을 때가 있습니다. Python에서는 matplotlib 라이브러리를 사용하여 손쉽게 그래프를 그릴 수 있습니다. 이 라이브러리는 다양한 형태의 그래프를 지원하며, 아래에서 몇 가지를 살펴보겠습니다. 더 많은 예시는 <a href="https://matplotlib.org/stable/tutorials/introductory/pyplot.html" target="_blank">여기</a>에서 확인하실 수 있습니다.

### 히스토그램

히스토그램은 데이터 집합에서 특정 값이나 특정 구간의 값이 얼마나 자주 나타나는지 보여 줍니다.

```python
import matplotlib.pyplot as plt # matplotlib 라이브러리 불러오기

# 정규 분포를 따르는 데이터 집합 생성
data = np.random.normal(0,5,1000)
plt.figure(num='histfig')   # 새 그림 창 생성

plt.hist(data,bins='auto',density=1, label='Occurence') # 데이터 그리기

plt.xlabel('Values')        # x축 레이블
plt.ylabel('Probability')   # y축 레이블
plt.legend()                # 범례 포함
plt.show()                  # 그래프 출력
```

### 산점도

데이터 집합의 측정값들은 산점도(scatter plot)로 나타낼 수 있습니다. 여기서는 점들만 표시하고(여기서는 검은색 점 `k.`) 점 사이를 선으로 잇지 않습니다. 아래는 산점도를 그리는 예시입니다. 


```python
import matplotlib.pyplot as plt
x = [1, 2, 3, 4, 5]
y = [1.1, 4.3, 9.2, 16.1, 25.1]

plt.figure(num='figscatterplot')

plt.plot(x,y,'k.', label='Measurements ')

# 시각적 요소 다듬기
plt.xlabel('Input voltage [V]')
plt.ylabel('Output voltage [V]')
plt.legend()
plt.grid()
plt.xlim(0,6)

plt.show()

```

## 곡선 맞춤

데이터에 가장 잘 맞는 곡선(최적 맞춤선)을 찾으려면 **scipy** 라이브러리의 `curve_fit` 함수를 사용할 수 있습니다(최소제곱 회귀를 사용합니다). 이 함수는 맞추려는 함수, $x$값, $y$값의 세 가지 입력을 필요로 하며, $x$와 $y$값은 측정 데이터가 담긴 배열입니다.

`curve_fit`은 `values`와 `covariance`라는 두 변수를 반환합니다. `values`는 맞춤 함수에 대한 최적의 매개변수 값들이 담긴 배열입니다.

`covariance`는 맞춤 매개변수들의 공분산을 담은 2차원 배열로, 대각선 원소들은 각 맞춤 매개변수의 분산을 나타냅니다. 맞춤 매개변수의 표준오차는 `np.sqrt(np.diag(covariance))`로 계산할 수 있습니다. (참고: 이 장은 Python의 핵심만 빠르게 훑어보는 안내입니다. 데이터 분석, 측정 불확도, Python에 대한 더 자세한 정보는 웹에서 쉽게 찾아보실 수 있습니다.)

```python
from scipy.optimize import curve_fit

def fit_function(x,a,b):
        return x**a+b

values, covariance = curve_fit(fit_function, x, y)
```

```python
x_test = np.linspace(0.8*min(x),1.2*max(x),1000) # x값을 위한 테스트 배열 생성
y_fit = fit_function(x_test,values[0],values[1]) # 테스트 배열에 대한 y값 계산

plt.figure(num='figscatterplot')

plt.plot(x,y,'k.', label='Measurements ')
plt.plot(x_test,y_fit,'r--', label='Fit')

# 시각적 요소 다듬기
plt.xlabel('Input voltage [V]')
plt.ylabel('Output voltage [V]')
plt.legend()
plt.grid()
plt.xlim(0,6)

plt.show()

print("a = ",values[0])
print("b = ",values[1])
```
