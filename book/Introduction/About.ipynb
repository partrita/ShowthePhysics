````python
# 라이브러리 불러오기
import numpy as np
import matplotlib.pyplot as plt

# 예시 데이터
x = np.array([1, 2, 3, 4, 5])   # x의 값들
y = 3.14*x**2                   # y의 값들

# 그래프 생성
plt.figure()
plt.plot(x, y, 'k.')            # 데이터 점을 검은색(k) 점(.)으로 그리기
plt.plot(x,y, 'r--')            # 데이터 점을 빨간색(r) 점선(--)으로 그리기

plt.xlabel('$x$[m]')            # 축 레이블 추가
plt.ylabel('$y$[s]')

plt.grid()                      # 격자 추가
plt.savefig('plot.png')         # 그래프 저장
plt.show()                      # 그래프 출력
````
