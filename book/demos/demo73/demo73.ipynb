## 물리학적 배경
모든 측정은 일정한 정밀도로만 수행할 수 있다. 정밀도는 측정 장비에 의해 결정되기도 하고, 측정하려는 물리량의 특성에 의해 결정되기도 한다. 과학에서는 측정 결과를 항상 관련 불확도와 함께 제시하며, 이때 반복 측정값이 측정값과 불확도 범위 안에 들어올 확률은 68%에 해당한다($P(\bar{x}-\mu_x<x<\bar{x}+\mu_x)=0.68$). 중력 가속도를 구할 때처럼 여러 물리량이 관여하는 경우, 각 측정의 불확도가 모두 결과에 영향을 미친다. 이 식에서 최종 불확도는 다음과 같이 주어진다:

$$
    (\frac{\mu_g}{g})^2 =  (\frac{\mu_H}{H})^2 + 4 (\frac{\mu_{\Delta t}}{\Delta t})^2
$$

불확도는 항상 유효숫자 한 자리로만 나타낸다. 최종 답은 불확도와 같은 소수점 자릿수까지 맞춘다(예: 9.8 ± 0.3 m/s²).
