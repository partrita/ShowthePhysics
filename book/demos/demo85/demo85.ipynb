## 소개
이 놀라운 시범 실험은 전구의 PTC(정온도 계수, Positive Temperature Coefficient) 특성을 눈으로 직접 확인할 수 있게 해 줍니다. 관찰한 현상을 설명하려면 많은 [오가며 사고하기](../../Pedagogy/BackAndForthThinking.md)가 필요합니다. 이 시범 실험은 다양한 수준의 학생들에게 흥미와 영감을 줄 수 있으며, 특히 직렬회로에서의 저항, 전류, 전압에 대한 학생들의 이해도를 확인하는 데 매우 적합합니다.

## 준비물
- 전원 장치
- 연결 도선을 포함한 동일한 전구 4~5개
- (선택 사항) 열화상(FLIR) 카메라

## 사전 준비
전구들이 직렬로 연결되어 있음이 분명히 보이도록 전구 사이의 도선을 미리 연결해 둡니다. 칠판에도 회로를 그려 둡니다.

## 실험 과정
먼저 전압을 일정하게 유지한 상태에서 전원에 연결되는 전구의 개수를 한 개씩 늘려 가며 시범 실험을 진행합니다. 학생들은 관찰 내용을 기록합니다. 그런 다음 관찰 결과를 간단히 정리합니다. 학생들에게서 나올 수 있는 관찰 내용은 다음과 같습니다.
- 연결된 두 단자 사이에 있는 전구만 켜진다.
- 연결된 전구가 많아질수록 불빛이 어두워진다.
- 마지막에 연결한 전구는 서서히 밝아지는 반면, 이미 연결되어 있던 다른 전구들은 즉시 최종 밝기로 켜진다.
- 연결하는 전구가 많아질수록 마지막에 연결한 전구가 최종 밝기에 도달하는 데 걸리는 시간이 점점 더 길어진다.

```{figure} demo85_figure2.JPG

연결을 옮긴 직후의 모습. 세 번째 전구는 거의 보이지 않을 정도로 희미하며, 다른 두 전구만큼 밝아지기까지 시간이 조금 걸린다.
```

이제 관찰한 현상에 대해 가능한 설명을 이끌어 내는 토론을 시작합니다. 세 번째와 네 번째 관찰 결과가 가장 흥미로우며, 첫 번째와 두 번째 관찰에 대한 설명은 학생들의 선행 지식을 확인하는 데 활용할 수 있습니다. 이러한 관찰을 모두 설명하려면 오가며 사고하기가 필요하며, 사고를 자극하는 질문을 던지고 학생들이 과학적 언어를 사용하여 설명하도록 유도하는 것이 중요합니다.

```{tip}
세 번째 관찰을 설명하도록 유도할 때는 전구에 불이 켜질 때 전구의 저항이 어떻게 변하는지 질문해 보세요. 만약 네 번째 관찰을 아직 하지 않았다면, 예측해 볼 수 있는 좋은 기회가 됩니다. 또한 마지막으로 연결한 전구가 천천히 밝아지는 동안 전류가 어떻게 변할지 예측하게 할 수도 있습니다. 전류는 감소하며, 이는 전류계로 확인할 수 있습니다.
```

## 물리학적 배경
세 번째 전구를 막 연결한 {numref}`Figure {number} <demo85_2>`의 상황에 세 번째와 네 번째 관찰을 연결하여 설명하는 것이 가장 좋습니다. 전구 1과 2는 이미 불이 켜져 있었고 전구 3은 그렇지 않았기 때문에, 전구 1과 2의 온도가 전구 3보다 높습니다. 따라서 전구 1과 2의 저항이 전구 3의 저항보다 크므로, 전구 1과 2에 더 큰 전압이 걸리게 됩니다.

연결된 후에는 전구 3의 온도(그리고 이에 따른 저항)가 상승하므로 전구 3에 점점 더 많은 전압이 걸리게 됩니다(반면 전구 1과 2에 걸리는 전압은 줄어들지만, 밝기 감소는 육안으로 구분되지 않을 정도입니다). 네 번째 관찰의 경우, 직렬로 연결된 전구가 많아질수록 전체 저항이 커져 전류가 작아지므로, 마지막 전구의 온도가 더 천천히 올라갑니다. 그 결과 마지막 전구에 걸리는 전압과 밝기도 더 느리게 증가합니다. 마지막으로 전류의 변화를 보면, 마지막으로 연결된 전구의 저항이 증가하면서 회로의 전체 저항이 커지므로 전류는 감소합니다.

## 후속 활동
이 시범 실험은 센서나 오실로스코프를 사용할 수 있을 때 심화 탐구로 발전시키기에 매우 좋습니다. 아래 사진은 옴 저항과 전구를 직렬로 연결한 회로에 교류(AC) 신호를 인가했을 때 나타나는 [이력 현상(hysteresis)](https://en.wikipedia.org/wiki/Hysteresis)을 보여 줍니다. {cite:t}`mooldijk2005`의 설명을 바탕으로 이를 구현한 시뮬레이션을 아래에 준비했습니다. 오른쪽 상단의 {fa}`rocket` 아이콘을 클릭해 실행해 보세요.

``` {figure} demo85_figure3.png
---
width: 50%
---
오실로스코프로 관측한 이력 현상. 신호 1은 5 Hz로 설정된 함수 발생기의 출력이고, 신호 2는 전구와 직렬로 연결된 저항 양단에서 측정한 전압입니다. 오실로스코프는 (X-Y) 모드로 설정되어 있습니다.
```

````python
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import interact, FloatSlider

# Model made by Ad Mooldijk

# Constants
U0 = 8          # V
Tg0 = 293      # K
l = 0.0143      # m
d = 0.132e-4    # m
pi = np.pi
A = pi * (d / 2)**2  # m^2
S = l * 2 * pi * (d / 2)  # m^2
Seff = S          # m^2
rho0 = 5.5e-8     # Ohm m
c = 0.13e3        # J/kg/K
rhod = 19.3e3     # kg/m^3
m = rhod * l * A  # kg
ssigma = 5.67e-8  # W/m^2/K^4
alpha = 4.9e-3    # K^-1
dt = float(1e-5)        # s

# Function to run simulation
def run_simulation(f):
    R = rho0 * l / A  # Initial resistance
    Tg = Tg0          # Initial temperature
    t = 0             # Initial time
    times = []
    temperatures = []
    U_x = []
    I_x = []

    # Simulation time parameters
    simulation_time = 3/f  # Total simulation time in seconds
    steps = int(simulation_time / dt)

    for _ in range(steps):
        U = U0 * np.sqrt(2) * np.sin(2 * pi * f * t)
        I = U / R
        Pin = I**2 * R
        Qin = Pin * dt
        dTgin = Qin / (c * m)
        Tg += dTgin
        dPuit = ssigma * Seff * Tg**4
        dQuit = dPuit * dt
        dTguit = -dQuit / (c * m)
        Tg += dTguit
        rho = rho0 * (1 + alpha * (Tg - Tg0))
        R = rho * l / A
        t += dt
        
        # Storing time, temperature, voltage, and current for plotting
        times.append(t)
        temperatures.append(Tg)
        U_x.append(U)
        I_x.append(I)

    # Plotting the results with subplots
    fig, ax = plt.subplots(2, 2, figsize=(12, 6))
    
    # Temperature over time
    ax[0,0].plot(times[int(steps/3):], temperatures[int(steps/3):])
    ax[0,0].set_xlabel('Time (s)')
    ax[0,0].set_ylabel('Temperature (K)')
    ax[0,0].grid(True)
    
    # Voltage vs. Current
    ax[0,1].plot(I_x[int(steps/3):], U_x[int(steps/3):])
    ax[0,1].set_xlabel('Current (A)')
    ax[0,1].set_ylabel('Voltage (V)')
    ax[0,1].grid(True)
    
    ax[1,0].plot(times[int(steps/3):], U_x[int(steps/3):])
    ax[1,0].set_xlabel('Time (s)')
    ax[1,0].set_ylabel('Voltage (V)')
    ax[1,0].grid(True)
    
    plt.tight_layout()
    plt.show()

# Slider for frequency
interact(run_simulation, f=FloatSlider(value=10, min=1, max=50, step=1, description='f in (Hz)'));
````
