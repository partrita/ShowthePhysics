## 물리학적 배경

정상파에는 다음 공식이 적용된다:
$v=\lambda f$

여기서 $v$는 파동의 속력(m/s), $\lambda$는 파장(m), $f$는 진동수(Hz)를 나타낸다. 팽팽한 줄에서의 파동의 속력은 다음 식으로도 계산할 수 있다:
$v=\sqrt{\frac{F_s}{\rho_L}}$

이 식에서 $v$는 마찬가지로 파동의 속력(m/s), $F_s$는 줄의 장력(N), $\rho_L$은 줄의 선밀도(kg/m)를 나타낸다. 선밀도는 줄의 질량을 줄의 길이로 나누어 구할 수 있다. 두 식을 같다고 두면 다음과 같다:
$\sqrt{\frac{F_s}{\rho_L}}=\lambda f$

이 실험에서의 변인은 파장과 줄의 장력이다. 실험에서 이 두 물리량 사이의 관계를 나타내면 위 식을 다음과 같이 정리할 수 있다:
$\lambda^2=\frac{F_s}{\rho_L f^2}$

그래프에서 줄의 장력에 대해 파장의 제곱을 나타내면 비례 관계가 성립한다. 그래프의 기울기는 $\frac{1}{\rho_L f^2}$와 같으며, 이로부터 진동수를 계산할 수 있다. 그래프의 예시는 {numref}`그림 {number} <demo77_fig2>`에 나와 있다.

## 후속 활동
선밀도가 다른 고무줄이나 다른 모델 또는 브랜드의 전동 칫솔을 사용하여 추가 탐구 실험을 진행해 볼 수 있다.


````python
# 필요한 라이브러리 불러오기
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

# 측정값 입력
mass = np.array([..,..,..,..,..]) # 추의 질량
wavelength = np.array([..,..,..,..,..]) # 측정한 파장

# 측정값 그래프 그리기
plt.figure
plt.plot(mass, wavelength, 'o')
plt.xlabel('mass')
plt.ylabel('wavelength')

# 축 범위 설정
plt.xlim(0,1.1*max(mass))
plt.ylim(0,1.1*max(wavelength))
plt.grid()
plt.show()
````

````python
# 데이터 피팅
def function(mass,a):
    return a*np.sqrt(mass*9.81)

var, cov = curve_fit(function, mass, wavelength)

print('The value of a is:', var[0])

# 측정 데이터와 피팅 결과 그래프 그리기
m_test = np.linspace(0.9*min(mass), 1.1*max(mass), 100)
w_test = function(m_test, var[0])

plt.figure
plt.plot(mass, wavelength, 'o')
plt.plot(m_test, w_test,'r--')
plt.xlabel('mass')
plt.ylabel('wavelength')

plt.xlim(0,1.1*max(mass))
plt.ylim(0,1.1*max(wavelength))
plt.grid()
plt.show()
````
