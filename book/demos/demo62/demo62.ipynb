## 물리학적 배경
LiDAR는 레이저 펄스를 이용해 물체나 표면까지의 거리를 측정하는 기술로, 레이더(radar)와 같은 원리로 작동한다. LiDAR 센서는 펄스를 방출한 시점부터 반사된 펄스를 수신할 때까지 걸린 시간을 측정하여 물체나 표면까지의 거리를 구한다: 

$$ s = \frac{c \Delta t}{2} $$

여기서 $s$는 거리(m), $c$는 빛의 속력(m/s), $\Delta t$는 펄스를 보내고 받을 때까지의 시간이다. 레이저 펄스는 물속에서 더 느리게 진행한다:

$$ n_{a\rightarrow w} = \frac{c_{air}}{c_{water}} $$ 

이처럼 이동 시간이 길어지기 때문에 LiDAR 센서에서 냄비 바닥까지의 거리가 '오차가 포함된' 거리(S4)로 측정된다. 물속 빛의 속력은 레이저 빛이 물속에 머무른 시간을 이용하여 계산한다({cite}`Bewersdorff2022` 참조):

$$ t_{water} = t_{measured} - t_{air} $$

$$ t_{water} = \frac{S4}{c_{air}} - \frac{S2}{c_{air}} $$

$$ c_{water} = \frac{S3}{t_{water}} $$ 

측정 조건에 따라 $c_{water}$의 문헌값($225\cdot10^6$ m/s)과 약 0.5~10%의 차이를 보인다.

````python
import numpy as np

# 단일 측정으로 물속 빛의 속력 구하기

# 상수
c_air = 299702547 # m/s, 공기 중 빛의 속력

# 측정값 (단위: m)
s2 = 0.29
s3 = 0.01
s4 = 0.3

# 계산
t_water = s4/c_air - s2/c_air
c_water = s3/t_water

# 결과를 보기 쉽게 출력
print("The speed of light in water is:", c_water*1e-8, "*10^8 m/s")
````

## Python 데이터 분석 예제
아래 코드 셀은 Norbert van Veen이 수집한 일부 측정값을 바탕으로 전체 데이터 분석을 수행하는 예시이다. (예를 들어 물이 없을 때 측정한 거리 등) 측정값 중 하나에 작은 측정 오차가 포함되기 쉽기 때문에 계통 오차(systematic error)를 도입하여 분석하였다.

````python
import numpy as np
import matplotlib.pyplot as plt
from scipy.optimize import curve_fit

# 여러 번의 측정값을 이용한 예시

# 상수
c_air = 299702547 # m/s, 공기 중 빛의 속력

# 측정값 (단위: m)
s1 = np.array([30,30,60.2,60.2,60.2,60.2,30,30,40,40,40,40,65.1,66])*1e-2
s4 = np.array([30.2,30.5,61.2,61.8,62.8,63.85,31.1,31.8,40.1,40.4,41.3,41.9,67.4,68])*1e-2
s2 = np.array([29,28,53.5,52.5,51.5,57.3,26,24,39,38,36,34,60,57.7])*1e-2

# 계산
s3 = s1 - s2
t_water = (s4-s2)/c_air
c_water = s3/t_water

# 물속 빛의 속력 배열 출력 - 사용하려면 주석을 해제하세요
#print(c_water)

# curve_fit을 사용하여 물속 빛의 속력과 가능한 계통 오차 구하기
def c_w(t,c_w,sys_err):
    return c_w*t + sys_err

var, cov = curve_fit(c_w,t_water,s3)

# 피팅 결과 값 출력
print('the systematic error is:', var[1]) 
print('the calculated value is:', var[0]) 


# 측정값과 피팅된 값을 함께 그리기 위한 배열 정의
x_test = np.linspace(min(t_water),max(t_water),1000)
y_test = c_w(x_test,*var)

# t_water에 따른 물의 깊이 그래프
plt.figure()

plt.plot(s3,t_water,'k.')
plt.plot(y_test,x_test,'r--')

plt.xlabel('$waterdepth$ (m)')
plt.ylabel('$t_{water}$ (s)')
plt.grid()

plt.show()

print('The speed of light in water is:',round(var[0]*1e-8,1),'+/-',round(np.sqrt(cov[0,0])*1e-8,1),'*10^8 m/s')
````
