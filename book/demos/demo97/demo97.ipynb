## 소개
방사성 붕괴는 통계적인 과정입니다. 학생들은 방사성 붕괴 과정을 직관적으로 떠올리기 어려워하므로, 이 시범 실험은 학생들이 이를 시각적으로 이해하는 데 도움을 줍니다.

한 변의 길이가 약 1.5 cm인 정육면체를 아주 많이(예: 300개) 준비합니다. 각 정육면체는 한 면을 파란색으로, 두 면을 빨간색으로 칠하고 나머지 세 면은 칠하지 않은 채로 둡니다. 이렇게 색칠한 정육면체를 던짐으로써 방사성 붕괴의 통계적 성질을 모의실험할 수 있습니다. 정육면체를 주사위처럼 책상 위에 던지면, 약 1/6은 파란색 면이, 2/6는 빨간색 면이, 3/6은 칠하지 않은(흰) 면이 위로 향하게 됩니다.


빨간색 면이 위로 나온 정육면체를 치우고, 나머지 정육면체를 다시 책상 위에 던집니다. 다시 빨간색 면이 위로 나온 정육면체를 빼내고 나머지를 던지는 과정을, 정육면체가 거의 남지 않을 때까지 반복합니다.

대신 파란색 면이 위로 나온 정육면체를 치울 수도 있습니다. 이 경우에는 모든 정육면체가 제거될 때까지 더 많은 횟수를 던져야 합니다.

## 준비물
통계적으로 잘 근사하려면 처음에 정육면체 300개 정도면 충분합니다.

## 사전 준비
단면이 정사각형인 긴 각목을 이용해 정육면체 300개를 만듭니다. 각목의 한 옆면은 파란색으로, 두 옆면은 빨간색으로 칠한 뒤 톱으로 잘라 내면, 정육면체의 나머지 세 면은 칠해지지 않은 상태로 남습니다.

## 실험 과정
* 처음 정육면체의 개수를 세고, 이 수를 $N(0)$으로 나타냅니다. 문자 $N$은 남은 정육면체의 수(Number)를, 숫자 0은 0번 던졌음을 의미합니다.
* 모든 정육면체를 처음으로 던집니다. 빨간색 면이 위로 나온 정육면체를 모두 치웁니다. 남은 정육면체의 수를 $N(1)$이라고 합니다.
이는 한 번 던진 후 남은 정육면체의 수입니다.
남은 정육면체를 두 번째로 던지고, 다시 빨간색이 나온 정육면체를 치웁니다. 이제 남은 수는 $N(2)$입니다.
그전에 정육면체가 다 떨어지지 않는 한 이와 같은 과정을 약 10회 반복합니다.
* 던진 횟수에 따른 남은 정육면체 수의 그래프, 즉 $n$에 대한 $N(n)$의 그래프를 그립니다.
* $N(n)$과 처음 개수 $N(0)$ 사이의 관계를 나타내는 수식을 세워 봅니다.
* 이 수식에 해당하는 그래프를 그립니다.
* 직접 세어서 그린 그래프와 수식으로 그린 그래프가 얼마나 잘 일치하는지 설명해 봅니다.

* 빨간색 면 대신 파란색 면이 위로 나온 정육면체를 치울 경우 그래프가 어떻게 달라질지 예측해 봅니다.
* 실제로 개수를 세어 보고, 이론적인 그래프와 얼마나 잘 맞는지 확인합니다.

시간에 따른 방사성 원자핵의 개수(또는 방사성 물질의 방사능 세기) 그래프를 보여 줍니다. 이 그래프는 다음 수식으로 나타낼 수 있습니다.
$N(t)=N(0) (1/2)^{(t/τ)}$
즉, 다음과 같은 의미를 가집니다.
*이 그래프의 모양은 남은 정육면체의 개수를 세어 그린 그래프의 모양과 매우 비슷합니다.*

## 물리학적 배경
* *정육면체*의 경우, 남아 있는 정육면체의 수는 처음 시작한 개수와 빨간색 면이 나올 확률에 따라 달라집니다.
* *원자핵*의 경우, 남아 있는 원자핵의 수는 처음 시작한 개수와 원자핵이 붕괴할 확률에 따라 달라집니다.

## 후속 활동
방사성 붕괴 과정에서는 방사성 물질 **A**가 물질 **B**로 붕괴합니다. 앞서 정육면체(물질 **A**)를 던지고 빨간색이 나온 정육면체(물질 **B**)를 치움으로써 이를 모의실험했습니다.
물질 **B** 역시 방사성이어서 물질 **C**로 붕괴하는 경우도 충분히 있을 수 있습니다. 즉 **A** → **B** → **C**의 과정을 거칩니다. 그러면 한편으로는 **A**의 붕괴로 인해 **B** 원자핵의 수가 증가하고, 다른 한편으로는 **C**로의 붕괴로 인해 **B** 원자핵의 수가 감소합니다. 물질 **C**가 안정하다면 **C**의 양은 계속 증가하기만 합니다.

두 종류의 던지기를 동시에 진행하면 이러한 상황도 모의실험할 수 있습니다. 첫 번째 던지기는 위에서 설명한 연속 과정이며, 두 번째 던지기는 다음과 같이 진행합니다.
* 치워 둔 빨간색 정육면체(물질 **B**)를 모았다가, 이것들도 함께 던집니다.
* 던졌을 때 파란색 면이 위로 나오면(물질 **C**) 치워 둡니다.
* 치워 둔 빨간색 정육면체의 수는 물질 **A**의 붕괴에 의해 늘어나지만, 던졌을 때 파란색이 나오면(물질 **C**) 줄어듭니다.
* 파란색 정육면체의 수는 계속 증가하기만 합니다.

```{tip}
이 시범 실험은 주사위로도 진행할 수 있습니다. 6면체보다 면의 수가 적은 주사위도 있습니다.

엑셀(Excel) 등을 이용하면 이러한 ‘붕괴 과정’을 컴퓨터 시뮬레이션으로 꽤 쉽게 구현할 수 있습니다. 아래에 파이썬(Python) 시뮬레이션을 제공해 두었습니다. 화면 상단의 로켓 아이콘을 클릭하면 직접 코드를 실행하고 수정해 볼 수 있습니다.
```



````python
# loading required libraries
import numpy as np
import matplotlib.pyplot as plt

# setting initial conditions
t_run = 1000 #s
dt = 1 #s

# half-life nuclei A
tau_A = 10 #s

# half-life nuclei B
tau_B = 30 #s

# Number of nucleus A
N_A_0 = 15000


# No need to change this code
N_total = int(t_run/dt)
N_A = np.zeros(N_total)
N_A[0] = N_A_0
N_B = np.zeros(N_total)
N_C = np.zeros(N_total)

t = np.zeros(N_total)

L_A = np.log(2)/tau_A
L_B = np.log(2)/tau_B


# calculating number of nuclei left
for i in range(len(N_A)-1):
    A_A = N_A[i]*L_A*np.exp(-L_A*dt)
    N_A[i+1] = N_A[i] - A_A*dt

    A_B = N_B[i]*L_B*np.exp(-L_B*dt)
    N_B[i+1] = N_B[i] - A_B*dt + A_A*dt

    N_C[i+1] = N_C[i] + A_B*dt

    t[i+1] = t[i] + dt

# plotting the data
plt.figure()
plt.plot(t,N_A,'k.',label='N_A')
plt.plot(t,N_B,'r.',label='N_B')
plt.plot(t,N_C,'b.',label='N_C')
plt.xlabel('time')
plt.ylabel('nuclei')
plt.legend()
plt.show()
````
